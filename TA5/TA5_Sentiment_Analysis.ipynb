```python
# IMPORT CELL
# Run this cell first to install and import the required libraries.

!pip install -q textblob nltk

import re
import nltk

from nltk.corpus import opinion_lexicon
from textblob import TextBlob

# Download the public word dataset required by the challenge.
nltk.download("opinion_lexicon", quiet=True)
```

```python
# ACTIVITY CELL
# Analyze 10 sentences using TextBlob and summarize their classifications.

positive_count = 0
neutral_count = 0
negative_count = 0

for number in range(1, 11):
    sentence = input(f"Enter sentence {number}: ").strip()

    # TextBlob returns polarity from -1 (negative) to 1 (positive).
    polarity = TextBlob(sentence).sentiment.polarity

    if polarity < 0:
        sentiment = "Negative"
        negative_count += 1
    elif polarity == 0:
        sentiment = "Neutral"
        neutral_count += 1
    else:
        sentiment = "Positive"
        positive_count += 1

    print(f"Sentiment: {sentiment}")
    print(f"Polarity: {polarity:.2f}\n")

# Display the number of sentences under each classification.
print("SUMMARY")
print(f"Positive sentences: {positive_count}")
print(f"Neutral sentences:  {neutral_count}")
print(f"Negative sentences: {negative_count}")
```

```python
# CHALLENGE CELL
# Use NLTK's public Opinion Lexicon as the happy- and sad-word datasets.

happy_words = set(opinion_lexicon.positive())
sad_words = set(opinion_lexicon.negative())


def extract_words(sentence):
    # Convert the sentence to lowercase words and remove punctuation.
    return re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?", sentence.lower())


def number_happy(sentence):
    # Count words found in the positive/happy-word dataset.
    return sum(word in happy_words for word in extract_words(sentence))


def number_sad(sentence):
    # Count words found in the negative/sad-word dataset.
    return sum(word in sad_words for word in extract_words(sentence))


sentence = input("Enter a sentence: ").strip()

happy_count = number_happy(sentence)
sad_count = number_sad(sentence)

# Determine the tone by comparing the two word counts.
if happy_count > sad_count:
    tone = "Positive"
elif sad_count > happy_count:
    tone = "Negative"
else:
    tone = "Neutral"

print(f"Happy words: {happy_count}")
print(f"Sad words: {sad_count}")
print(f"Analysis: {tone}")
```
