# Rule Based Vacuum Cleaner Agent

This notebook implements the two-room rule-based agent from Technical Assessment 2. It creates the environment, accepts room states and a step count, applies the cleaning and movement rules, and prints the result after every step.

## Task 1 Set Up the Environment

The environment is a dictionary containing rooms A and B. The helper functions validate room states, check whether a room is dirty, and clean it.

```python
def normalize_state(state):
    """Return 'Dirty' or 'Clean'; reject any other room state."""
    normalized = state.strip().capitalize()
    if normalized not in {'Dirty', 'Clean'}:
        raise ValueError("Room state must be 'dirty' or 'clean'.")
    return normalized

def create_environment(room_a_state, room_b_state):
    """Initialize the two rooms using a Python dictionary."""
    return {
        'A': normalize_state(room_a_state),
        'B': normalize_state(room_b_state),
    }

def is_dirty(environment, room):
    """Check whether the selected room is dirty."""
    return environment[room] == 'Dirty'

def clean_room(environment, room):
    """Change the selected room's state to Clean."""
    environment[room] = 'Clean'
```

## Task 2 Implement the Rule Based Agent

Rule 1 cleans the current room when it is dirty. Rule 2 moves the agent to the other room when the current room is already clean.

```python
class VacuumAgent:
    def __init__(self, environment, starting_location='A'):
        if starting_location not in environment:
            raise ValueError('Starting location must exist in the environment.')
        self.environment = environment
        self.location = starting_location

    def move(self):
        """Switch between rooms A and B."""
        self.location = 'B' if self.location == 'A' else 'A'
        return f'Moving to room {self.location}'

    def perceive_and_act(self):
        """Inspect the current room, then clean it or move."""
        if is_dirty(self.environment, self.location):
            room_cleaned = self.location
            clean_room(self.environment, room_cleaned)
            return f'Room {room_cleaned} has been cleaned.'
        return self.move()
```

## Task 3 Run the Simulation

`run_interactive_simulation()` asks for both initial room states and the number of steps. Input is checked so only `dirty` or `clean` states and a positive whole-number step count are accepted.

```python
def ask_room_state(room):
    while True:
        state = input(f'Is room {room} Dirty or Clean? ')
        try:
            return normalize_state(state)
        except ValueError as error:
            print(error)

def ask_steps():
    while True:
        try:
            steps = int(input('How many steps should the agent run? '))
            if steps <= 0:
                raise ValueError
            return steps
        except ValueError:
            print('Please enter a positive whole number.')

def run_simulation(environment, steps, starting_location='A'):
    agent = VacuumAgent(environment, starting_location)
    for step in range(1, steps + 1):
        location_before_action = agent.location
        action = agent.perceive_and_act()
        print(f'Step {step}: Agent is in room {location_before_action}')
        print(action)
        print(f'Environment state: {environment}')
        if step < steps:
            print()
    return agent

def run_interactive_simulation():
    room_a_state = ask_room_state('A')
    room_b_state = ask_room_state('B')
    print()
    steps = ask_steps()
    print()
    environment = create_environment(room_a_state, room_b_state)
    return run_simulation(environment, steps)
```
